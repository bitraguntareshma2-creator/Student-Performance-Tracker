```python
# CODETECH Internship - Task 2
# Project: Student Performance Tracker

import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

data = {
    "Name": ["Priya", "Arun", "Sneha", "Rahul"],
    "Total": [273, 173, 225, 135],
    "Percentage": [91.00, 57.67, 75.00, 45.00],
    "Grade": ["A+", "D", "B", "F"],
    "Status": ["Pass", "Pass", "Pass", "Pass"]
}

df = pd.DataFrame(data)
display(df)

df.to_csv("student_performance_report.csv", index=False)

plt.bar(df["Name"], df["Total"])
plt.xlabel("Student Name")
plt.ylabel("Total Marks")
plt.title("Student Performance Tracker")
plt.show()
```
